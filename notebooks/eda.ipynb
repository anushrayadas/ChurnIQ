```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)
```

```python
data_path = r"C:\Users\anush\OneDrive\Desktop\churnIQ\data\processed\churn_processed.csv"

df = pd.read_csv(data_path)

print(df.shape)
df.head()
```

```python
print(df.info())

print("\nMissing values:\n")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(6,4))

ax = sns.countplot(x="Churn", data=df)

plt.title("Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Customer Count")

for p in ax.patches:
    ax.annotate(
        f'{int(p.get_height())}',
        (p.get_x() + p.get_width()/2., p.get_height()),
        ha='center',
        va='bottom'
    )

output_path = r"C:\Users\anush\OneDrive\Desktop\churnIQ\docs\screenshots\churn_distribution.png"

plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(x="Contract", hue="Churn", data=df)

plt.title("Contract Type vs Churn")
plt.xlabel("Contract Type")
plt.ylabel("Customer Count")
plt.xticks(rotation=15)

output_path = r"C:\Users\anush\OneDrive\Desktop\churnIQ\docs\screenshots\contract_vs_churn.png"

plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="Churn", y="tenure", data=df)

plt.title("Tenure vs Churn")
plt.xlabel("Churn")
plt.ylabel("Tenure (months)")

output_path = r"C:\Users\anush\OneDrive\Desktop\churnIQ\docs\screenshots\tenure_vs_churn.png"

plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="Churn", y="MonthlyCharges", data=df)

plt.title("Monthly Charges vs Churn")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")

output_path = r"C:\Users\anush\OneDrive\Desktop\churnIQ\docs\screenshots\monthly_charges_vs_churn.png"

plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.show()
```
